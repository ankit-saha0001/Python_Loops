### Task 1:- Discount Rules
#### In this task discount will be given in accordance with the price given 
#### Input error handling will also be taken care of along with tax calculation

```python
## Input of the price along with error handling
try:
    price = float(input("Enter the price of the product: "))
    print(price)
except ValueError:
    print("Please enter a valid number for the price.")
```

```python
## Discount will be given to the product as per the price

if price>=2000:
    discount=0.15
elif price>=1500 and price<2000:
    discount=0.10
elif price>=1000 and price<1500:
    discount=0.07
else :
    discount=0

discount_price=price-price*discount
print(f"The subtotal after considering the offer available is: {discount_price}")

tax=0.05
total_price=discount_price*(1+tax)
print(f"The total price after adding the tax is: {total_price}")
```

### Task 2:- Process Multiple Orders 
#### In this task multiple order will get process using for loop and total revenue will be calculated

```python
## Processing multiple orders and calculating the total price for each order
orders=[1200,2500,800,1750,3000]
discounted_items=0
revenue=0
## The loop iterates through each product and gives the discount as per the price
for product in orders:
    if product>=2000:
        discount=0.15
        discounted_items+=1
    elif product>=1500 and product<2000:
        discount=0.10
        discounted_items+=1
    elif product>=1000 and product<1500:
        discount=0.07
        discounted_items+=1
    else :
        discount=0
    

    discount_price=product-product*discount
    revenue+=discount_price
    print(f"The product value is {product}, the discount applied is {int(discount*100)}%, and the price after discount is {discount_price}")

print(f"The total revenue from the orders is: {revenue}")
print(f"The number of discounted items is: {discounted_items}")   
```

## Task 3:- User Menu
#### Enabling user to add their order and update them of their order value after their discount


```python
message="Menu options:\n1. Add order amount to a running list of orders\n2. Show all orders and totals after applying discounts\nq. Quit to exit the program\n"

print(message)
order=[]
while True:
    choice =input("Enter your choice:")
    ## Adding order amount to the running list of orders and applying discounts
    if choice == '1':
        try:
            amount = float(input("Enter the order amount: "))
        except ValueError:
            print("Invalid input. Please enter a valid number.\n")
            print(message)
            continue

        if amount >= 2000:
            discount = 0.15
        if amount >= 1500 and amount < 2000:
            discount = 0.10
        if amount >= 1000 and amount < 1500:
            discount = 0.07
        if amount < 1000:
            discount = 0    
        discount_price=amount-amount*discount
        order.append(discount_price)
    ## Showing all orders after discount
    elif choice == '2':
        for amount in order:
            print(f"Price of order after discount: {amount}\n")
        print(f"Total amount of orders: {sum(order)}\n")
    ## Exiting the program
    elif choice=='q':
        print("Exiting the program.")
        break    
    ## Handling invalid choices
    else:
        print("Invalid choice. Please try again.\n")
        print(message)
        continue
```

## Task 4:- Loop Control with Conditions (break and continue)
#### Checking the daily sales and update the total sales 

```python
## List of daily sales
daily_sales=[200,150,0,400,50,-1,300]
## Calculating total sales and handling corrupted values
total_sales=0
days=0
for sale in daily_sales:
    days+=1
    if sale<0:
        print(f"Corrupted value found: {sale} on the {days}th day.\nBreaking the loop.")
        break
    if sale==0:
        print(f"No sales for {days}th day.")
        continue        
    total_sales+=sale

print(f"Total sales for the day before the corrupted value: {total_sales}")
```
